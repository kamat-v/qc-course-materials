**Week 1**: Complex Numbers and Linear Algebra Warm-Up

```python
import numpy as np
```

1. Complex Numbers: Cartesian / Polar Form

```python
z = 3 - 4j

# Cartesian form
print("z =", z)

# Magnitude and phase
r = np.abs(z)
theta = np.angle(z)

print("Magnitude |z| =", r)
print("Phase arg(z) =", theta)

# Reconstruct z from polar form
z_reconstructed = r * np.exp(1j * theta)
print("Reconstructed z =", z_reconstructed)
```

Vectors over $\mathbb{C}$, and inner products



```python
v = np.array([1 + 1j, 2 - 1j])
w = np.array([1 - 1j, 4+1j])

# Inner products (using bra-ket notation)
inner_vv = np.vdot(v, v)   # <v|v>
inner_vw = np.vdot(v, w)   # <v|w>
inner_wv = np.vdot(w, v)   # <w|v>

print("<v|v> =", inner_vv)
print("<v|w> =", inner_vw)
print("<w|v> =", inner_wv)
```

Matrix-Vector Multiplication and Adjoints


```python
A = np.array([[1, 1j],
              [-1j, 2]])

x = np.array([1, 1 + 1j])
Ax = A @ x

lhs = Ax.conj().T
rhs = x.conj().T @ A.conj().T

print("Real part of Ax dagger =", lhs.real)
print("Imaginary part of Ax dagger", lhs.imag)
print("Real part of x dagger A dagger =", rhs.real)
print("Imaginary part of x dagger A dagger=", rhs.imag)
```

Hermitian and Unitary Matrices

```python
H = np.array([[2, 1j],
              [-1j, 3]])
print("H =\n", H)
print("H dagger =\n", H.conj().T)
```

```python
theta = np.pi / 4

U = np.array([[np.exp(1j * theta), 0],
              [0, np.exp(-1j * theta)]])
print("U†U =\n", U.conj().T @ U)
print("U is unitary?", np.allclose(U.conj().T @ U, np.eye(2)))
# Norm preservation
Ux = U @ x
print("Norm of x =", np.linalg.norm(x))
print("Norm of Ux =", np.linalg.norm(Ux))
```
